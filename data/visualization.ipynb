# Airport Map Visualization

This project demonstrates how to download airport data from the X-Plane Scenery Gateway and create interactive airport maps.

## Why X-Plane Scenery Gateway?

The X-Plane scenery system provides a rich source of airport data (runways, taxiways, frequencies, lighting systems, etc.) in a simple `.dat` text format. These datasets are:

- **Free and easily accessible** via the official API
- **Useful for academic or personal projects**, such as visualizations, simulations, or research
- **Comprehensive data** including precise coordinates, elevations, surface types, and lighting information

## Limitations

⚠️ **Not suitable for real-world operational projects**: Updates depend on community contributors who modify airports in their free time, so the data may not always reflect real-time changes.

✅ **Best for**: Educational, simulation, or visualization purposes.

## File Format Overview

The airport data is stored in `apt.dat` files with **row codes** describing different elements:

| Row Code | Element Type | Description |
|----------|--------------|-------------|
| **1** | Airport Header | Basic airport information |
| **100** | Runway | Runway definitions with both ends |
| **110** | Pavement | Taxiways, ramps, aprons |
| **120** | Linear Feature | Roads, boundaries |
| **130** | Airport Boundary | Perimeter definition |
| **1200** | Taxi Route Network | Ground movement routes |
| **1300** | Startup Locations | Aircraft parking positions |

## Runway Format Example

Here's a complete runway definition from `apt.dat`:

```
100  29.87 1 0 0.15 0 2 1 13L 47.53801700 -122.30746100 73.15 0.00 31R 47.52068300 -122.30746100 73.15 0.00
```

### Field Breakdown:

| Field | Value | Description |
|-------|-------|-------------|
| **1** | `100` | **Row code**: indicates this line describes a runway |
| **2** | `29.87` | **Runway width** in meters |
| **3** | `1` | **Surface type** (1 = asphalt, 2 = concrete, 3 = grass, etc.) |
| **4** | `0` | **Water runway** indicator (0 = no, 1 = yes) |
| **5** | `0.15` | **Runway friction coefficient** |
| **6** | `0` | **Shoulder** (reinforced edge): 0 = none |
| **7** | `2` | **Lighting type** (0 = none, 1 = basic, 2 = standard, etc.) |
| **8** | `1` | **Runway markings** (0 = none, 1 = basic, 2 = non-precision, etc.) |

### Runway End 1 (13L):
| Field | Value | Description |
|-------|-------|-------------|
| **9** | `13L` | **Runway identifier** for first end |
| **10** | `47.53801700` | **Latitude** in decimal degrees |
| **11** | `-122.30746100` | **Longitude** in decimal degrees |
| **12** | `73.15` | **Elevation** in meters above sea level |
| **13** | `0.00` | **Threshold displacement** in meters |

### Runway End 2 (31R):
| Field | Value | Description |
|-------|-------|-------------|
| **14** | `31R` | **Runway identifier** for second end |
| **15** | `47.52068300` | **Latitude** in decimal degrees |
| **16** | `-122.30746100` | **Longitude** in decimal degrees |
| **17** | `73.15` | **Elevation** in meters above sea level |
| **18** | `0.00` | **Threshold displacement** in meters |

> **Note**: Runway ends are always opposite (approximately 180° difference). In this example, 13L points toward ~130° and 31R toward ~310°.
